## 🧬 Load reference and reads data into variables
Below, we provide some helper code to load the reference from `reference.txt` and the reads from `reads.csv`.

❗ **Do not change the function’s behavior** (you may read it, of course!).

The code creates two variables from the loaded files:
- **reference**: a string containing the 200-nt reference sequence
- **reads**: a list of dictionaries containing the 6,000 reads

````python
##### Constants

REF_PATH = "reference.txt"
READS_PATH = "reads.csv"

##### Functions

def load_reference(path: str) -> str:
    """Load reference nucleotide sequence from file at path to reference string."""
    with open(path) as f:
        lines = [ln.strip() for ln in f if ln.strip()]
    if lines[0].startswith('>'):
        lines = lines[1:]
    ref = "".join(lines).upper()
    if len(ref) != 200:
        raise ValueError(f"Reference must be 200 nt, got {len(ref)}")
    return ref


def load_reads(path: str) -> list:
    """Load sample NGS reads from file at path to reads list."""
    with open(path, newline="") as f:
        reader = csv.DictReader(f)
        reads = list(reader)
    return reads

##### Main code

# Load reference and reads
reference = load_reference(REF_PATH)
reads = load_reads(READS_PATH)

# Report data structures
print("Reference length:", len(reference))
print("Total no. of reads:", len(reads))
print("Keys in each read dictionary:", reads[0].keys())
print("Samples loaded:", sorted({read["sample"] for read in reads}))

# Print sample sequences
print("\nSample read data:")
for row in reads[:5]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
print("...")
for row in reads[-5:]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
````

## Step 1 — Validate each read and record any errors
## Function: `validate_read(sequence, reference)`

### 👍 You can assume:
- `reference` is a non-empty string containing only uppercase A/C/G/T.
- `sequence` is a string, but nothing about its length or contents.

### 🎯 Aims:
- Check that an individual read is the same length as the reference and contains only valid characters (A/C/G/T)
- "Repair" any reads that contain valid lower-case characters -> upper-case
- Return a dictionary with the following keys:

    - `sequence`: the upper-case version of the sequence input, including when invalid.
    - `valid`: `True` only when the length matches the reference and every character is A/C/G/T.
    - `errors`: a list of strings. Include `"wrong_length"` first if lengths differ, then `"invalid_characters"` if any character is outside A/C/G/T. Include both when both apply. A valid read has `[]`.

### 🔎 Expected outputs

Examples for reference `"ATGC"`:
| Input (read) | Input type | Output |
|---|---|:---|
| `"ATGC"` | Correct | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `"atgc"` | Lowercase | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `""` | Empty read | `{"sequence": "", "valid": False, "errors": ["wrong_length"]}` |
| `"ATNC"` | Invalid character | `{"sequence": "ATNC", "valid": False, "errors": ["invalid_characters"]}` |
| `"ATN"` | Multiple errors | `{"sequence": "ATN", "valid": False, "errors": ["wrong_length", "invalid_characters"]}` |


Invalid reads should not crash your program.

### 💻 Enter your code here

````python
def validate_read(sequence, reference):

    """
    Normalise and validate a DNA sequencing read.

    Parameters:
        sequence (str): The input DNA sequencing read.
        reference (str): The reference DNA sequence.

    Returns:
        dict: The uppercase sequence, its validity status,
              and a list of any validation errors.
    """

    # Convert the sequence to uppercase
    cleaned_sequence = sequence.upper()

    # Create an empty list to store validation errors
    errors = []

    # Check whether the read is the same length as the reference
    if len(cleaned_sequence) != len(reference):
        errors.append("wrong_length")

    # Check for characters other than valid DNA nucleotides
    if any(nucleotide not in "ACGT" for nucleotide in cleaned_sequence):
        errors.append("invalid_characters")

    # Return the validation results as a dictionary
    return {
        "sequence": cleaned_sequence,
        "valid": not errors,
        "errors": errors
    }
````

## Step 2 — Find every mismatch in a valid read
## Function: `find_mutations(sequence, reference)`

### 👍 You can assume:
- Inputs `sequence` and `reference` are valid (i.e. both are strings of the same length and contain only upper-case bases, i.e. A/C/G/T)

### 🎯 Aims:
- Return a list containing one dictionary per mismatch, in ascending position order.
- Each dictionary has exactly these keys:
    - `position`: position of the mutation (integer), numbered from 1.
    - `reference_base`: one-character string from reference at that position.
    - `read_base`: one-character string from the read at that position.

- Return `[]` if the read matches the reference.

### 🔎 Expected outputs

Example: `find_mutations("GTGA", "ATGC")` returns:
```python
[{"position": 1, "reference_base": "A", "read_base": "G"},
 {"position": 4, "reference_base": "C", "read_base": "A"}]
```


### 💻 Enter your code here

````python
def find_mutations(sequence, reference):
    """
    Identify nucleotide mismatches between a sequencing read
    and a reference sequence.

    Parameters:
        sequence (str): A valid DNA sequencing read.
        reference (str): The reference DNA sequence.

    Returns:
        list: A list of dictionaries containing the position,
              reference base and read base of each mismatch.
              An empty list is returned if no mismatches exist.
    """

    # Create an empty list to store mutations
    mutations = []

    # Compare each nucleotide with the reference sequence
    for i in range(len(reference)):
        # Identify mismatches between the read and reference
        if sequence[i] != reference[i]:
            # Record the position and nucleotide substitution
            mutation = {
                "position": i + 1,
                "reference_base": reference[i],
                "read_base": sequence[i]
            }
            mutations.append(mutation)

    return mutations
````

### 👍 You can assume:
- `reads` is a list of dictionaries containing keys `read_id`, `sample`, `group` and `sequence`, each with string values.
- `reference` is a non-empty string containing only A/C/G/T.

```python
    ]

```
- Do not change the original read dictionaries.

### 🔎 Expected outputs
- Valid reads have `status="valid"` and `mutations` listing all mutations in the dictionary format shown above.
- Invalid reads have `status="invalid"`, the relevant error codes, and `mutations=[]`.
- Empty input returns `[]`.

### 💻 Enter your code here

````python
def analyse_reads(reads, reference):
    """
    Validate and analyse all sequencing reads for mutations.

    Parameters:
        reads (list): A list of dictionaries containing sequencing
                      reads and associated sample information.
        reference (str): The reference DNA sequence.

    Returns:
        list: A new list of dictionaries containing normalised
              sequences, validation status, errors and mutations.
              The original read dictionaries remain unchanged.
    """

    # Create an empty list to store analysed reads
    analysed_reads = []

    # Process each sequencing read individually
    for read in reads:

        # Validate and normalise the sequence using Step 1
        validation = validate_read(read["sequence"], reference)

        # Identify mutations only if the read is valid
        if validation["valid"]:
            status = "valid"
            mutations = find_mutations(
                validation["sequence"], reference
            )

        else:
            status = "invalid"
            mutations = []

        # Create a new dictionary containing the analysis results
        analysed_read = {
            "read_id": read["read_id"],
            "sample": read["sample"],
            "group": read["group"],
            "sequence": validation["sequence"],
            "status": status,
            "errors": validation["errors"],
            "mutations": mutations
        }

        # Add the analysed read to the results list
        analysed_reads.append(analysed_read)

    return analysed_reads
````

````python
# Write your code here.
def print_table(headers, rows):
    """
    Print a formatted table with automatically adjusted column widths.

    Parameters:
        headers (list): List of column headings.
        rows (list): List of rows containing string values.
    """

    table = [headers] + rows

    # Calculate the required width of each column
    widths = [
        max(len(row[i]) for row in table)
        for i in range(len(headers))
    ]

    # Print each row using f-string formatting
    for row in table:
        print(" | ".join(
            f"{value:<{width}}"
            for value, width in zip(row, widths)
        ))


def print_sample_summaries(summary):
    """
    Print mismatch summary and substitution mutation tables
    separately for each experimental group.

    Parameters:
        summary (dict): Dictionary of sample-level statistics.
    """

    # Identify all experimental groups
    groups = sorted({
        data["group"] for data in summary.values()
    })

    for group in groups:

        print(f"\nGroup: {group}")

        # Identify samples belonging to the current group
        samples = sorted(
            sample for sample in summary
            if summary[sample]["group"] == group
        )

        # TABLE 1: Summarised mismatch results
        print("\nMismatch summary")

        headers = [
            "Sample",
            "Valid reads",
            "Invalid reads",
            "Valid reads with ≥1 mismatch (%)",
            "Mismatch events",
            "Top positions"
        ]

        rows = []

        for sample in samples:
            data = summary[sample]
            percentage = data["mismatch_percentage"]

            # Handle samples containing no valid reads
            if percentage is None:
                percentage_text = "N/A"
            else:
                percentage_text = f"{percentage:.2f}%"

            positions = ", ".join(
                str(position)
                for position in data["top_positions"]
            )

            rows.append([
                sample,
                str(data["valid_reads"]),
                str(data["invalid_reads"]),
                percentage_text,
                str(data["mismatch_events"]),
                positions or "None"
            ])

        print_table(headers, rows)

        # TABLE 2: Substitution mutations
        print("\nSubstitution mutations (count and % per 100 valid reads)")

        # Identify all observed substitutions within the group
        substitutions = sorted({
            substitution
            for sample in samples
            for substitution in summary[sample]["substitution_counts"]
        })

        if not substitutions:
            print("No substitutions observed.")
            continue

        rows = []

        for substitution in substitutions:
            row = [substitution]

            for sample in samples:
                data = summary[sample]

                # Return zero if the substitution was not observed
                count = data["substitution_counts"].get(
                    substitution, 0
                )

                valid_reads = data["valid_reads"]

                # Calculate substitutions per 100 valid reads
                if valid_reads == 0:
                    row.append(f"{count} (N/A)")
                else:
                    percentage = 100 * count / valid_reads
                    row.append(f"{count} ({percentage:.2f}%)")

            rows.append(row)

        print_table(["Substitution"] + samples, rows)


def get_top_positions(counts, n=3):
    """
    Rank mismatch positions by descending frequency.
    Resolve ties by prioritising earlier positions.
    """

    return sorted(
        (position for position in counts if counts[position] > 0),
        key=lambda position: (-counts[position], position)
    )[:n]


def summarise_samples(analysed_reads):
    """
    Calculate sample-level mismatch statistics and print
    summary tables separately for each experimental group.

    Parameters:
        analysed_reads (list): List of annotated sequencing
                               read dictionaries from Step 3.

    Returns:
        dict: Nested dictionary containing mismatch statistics
              for each sample.

    The original analysed read dictionaries remain unchanged.
    """

    # Create an empty dictionary to store sample summaries
    summary = {}

    for read in analysed_reads:

        sample = read["sample"]

        # Initialise statistics when a new sample is encountered
        if sample not in summary:

            summary[sample] = {
                "group": read["group"],
                "valid_reads": 0,
                "invalid_reads": 0,
                "mismatch_reads": 0,
                "mismatch_percentage": None,
                "mismatch_events": 0,
                "substitution_counts": {},
                "position_counts": {},
                "top_positions": []
            }

        data = summary[sample]

        # Count invalid reads and exclude them from mismatch analysis
        if read["status"] == "invalid":
            data["invalid_reads"] += 1
            continue

        # Count valid reads
        data["valid_reads"] += 1

        mutations = read["mutations"]

        # Count reads containing at least one mismatch
        if mutations:
            data["mismatch_reads"] += 1

        # Count every individual mismatch event
        data["mismatch_events"] += len(mutations)

        # Record substitution types and mutation positions
        for mutation in mutations:

            substitution = (
                mutation["reference_base"] + ">" +
                mutation["read_base"]
            )

            position = mutation["position"]

            substitution_counts = data["substitution_counts"]
            position_counts = data["position_counts"]

            substitution_counts[substitution] = (
                substitution_counts.get(substitution, 0) + 1
            )

            position_counts[position] = (
                position_counts.get(position, 0) + 1
            )

    # Calculate percentages and identify the top mutation positions
    for data in summary.values():

        if data["valid_reads"] > 0:
            data["mismatch_percentage"] = (
                100 * data["mismatch_reads"] /
                data["valid_reads"]
            )

        # Identify top positions using the shared helper inside the loop
        data["top_positions"] = get_top_positions(
            data["position_counts"]
        )

    # Print the results separately for each experimental group
    print_sample_summaries(summary)

    return summary
````

## Function:
`compare_positions(sample_summary, control_group="control", treated_group="treatment")`

### 👍 You can assume

### 🎯 Aims

````python
#@title 📊 calculate_statistics function

from scipy import stats
import math
import warnings


def calculate_statistics(control_percentages, treated_percentages):
    """Compare independent biological samples using a two-sided Welch's t-test."""
    unavailable = {"statistic": None, "p_value": None}

    # Require at least two samples per group and finite numerical values.
    if len(control_percentages) < 2 or len(treated_percentages) < 2:
        return unavailable

    values = list(control_percentages) + list(treated_percentages)
    if any(value is None or not math.isfinite(value) for value in values):
        return unavailable

    # No within-group variation: the t-test cannot estimate uncertainty.
    if (
        len(set(control_percentages)) == 1
        and len(set(treated_percentages)) == 1
    ):
        return unavailable

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        result = stats.ttest_ind(
            treated_percentages,
            control_percentages,
            equal_var=False,
            alternative="two-sided"
        )

    if not (
        math.isfinite(result.statistic)
        and math.isfinite(result.pvalue)
    ):
        return unavailable

    return {
        "statistic": float(result.statistic),
        "p_value": float(result.pvalue)
    }
````

```python
    }
```

Print **one brief line per position**, reporting group means, their difference, sample numbers and p-value.

### 💻 Enter your code here

````python

from statistics import mean, variance
from math import isfinite


def welch_df(control_percentages, treated_percentages):
    """
    Calculate the degrees of freedom for Welch's t-test.

    Parameters:
        control_percentages (list): Control sample percentages.
        treated_percentages (list): Treated sample percentages.

    Returns:
        float or None: Degrees of freedom, or None if undefined.
    """

    # At least two biological samples per group are required
    if len(control_percentages) < 2 or len(treated_percentages) < 2:
        return None

    control_term = (
        variance(control_percentages) / len(control_percentages)
    )

    treated_term = (
        variance(treated_percentages) / len(treated_percentages)
    )

    denominator = (
        control_term**2 / (len(control_percentages) - 1)
        + treated_term**2 / (len(treated_percentages) - 1)
    )

    # Avoid division by zero when both groups have zero variance
    if denominator == 0:
        return None

    return (control_term + treated_term)**2 / denominator


def get_sample_percentages(samples, position, group):
    """
    Calculate mismatch percentages for each eligible sample
    belonging to the specified experimental group.

    Parameters:
        samples (dict): Samples containing valid reads.
        position (int): Reference nucleotide position.
        group (str): Experimental group to analyse.

    Returns:
        list: One mismatch percentage per biological sample.
    """

    return [
        100 * data["position_counts"].get(position, 0)
        / data["valid_reads"]
        for data in samples.values()
        if data["group"] == group
    ]


def print_position_result(result, df, control_group, treated_group):
    """
    Print a single line summarising the mismatch comparison.

    Parameters:
        result (dict): Statistical results for one position.
        df (float or None): Welch's degrees of freedom.
        control_group (str): Control group label.
        treated_group (str): Treatment group label.
    """

    control_mean = result["control_mean"]
    treated_mean = result["treated_mean"]
    difference = result["difference"]

    # Format means and differences, including unavailable values
    control_text = (
        f"{control_mean:.2f}%"
        if control_mean is not None else "N/A"
    )

    treated_text = (
        f"{treated_mean:.2f}%"
        if treated_mean is not None else "N/A"
    )

    difference_text = (
        f"{difference:+.2f} pp"
        if difference is not None else "N/A"
    )

    statistic = result["statistic"]
    p_value = result["p_value"]

    # Format the statistical results
    if statistic is None or p_value is None:
        test_text = "Welch's t-test unavailable; p = N/A"

    else:
        df_text = f"({df:.2f})" if df is not None else ""

        if p_value < 0.001:
            p_text = "p < 0.001"
        else:
            p_text = f"p = {p_value:.3f}"

        test_text = (
            f"two-sided Welch's t-test: "
            f"t{df_text} = {statistic:.3f}, {p_text}"
        )

    # Print one summary line per nucleotide position
    print(
        f"Position {result['position']}: "
        f"{control_group} mean = {control_text} "
        f"(n = {result['control_n']}), "
        f"{treated_group} mean = {treated_text} "
        f"(n = {result['treated_n']}), "
        f"difference = {difference_text}; "
        f"{test_text}"
    )


def compare_positions(
    sample_summary,
    control_group="control",
    treated_group="treatment"
):
    """
    Compare mismatch percentages between two experimental
    groups at the three most frequently mismatched positions.

    Parameters:
        sample_summary (dict): Sample-level results from Step 4.
        control_group (str): Control group label.
        treated_group (str): Treatment group label.

    Returns:
        list: Dictionaries containing the position, sample sizes,
              group means, difference, test statistic and p-value.
    """

    selected_samples = {}
    combined_counts = {}

    # Identify samples belonging to the two selected groups
    for sample, data in sorted(sample_summary.items()):

        if data["group"] not in (control_group, treated_group):
            continue

        # Exclude samples without valid reads
        if data["valid_reads"] == 0:
            print(
                f"Excluded {sample} ({data['group']}): "
                "no valid reads."
            )
            continue

        selected_samples[sample] = data

        # Combine mismatch counts across the selected groups
        for position, count in data["position_counts"].items():
            combined_counts[position] = (
                combined_counts.get(position, 0) + count
            )

    # Reuse the Step 4 helper to rank mismatch positions
    top_positions = get_top_positions(combined_counts)

    results = []

    # Handle datasets without mismatches
    if not top_positions:
        print("No mismatch positions found in the selected groups.")
        return results

    # Compare the experimental groups at each selected position
    for position in top_positions:

        control_percentages = get_sample_percentages(
            selected_samples, position, control_group
        )

        treated_percentages = get_sample_percentages(
            selected_samples, position, treated_group
        )

        # Calculate group means
        control_mean = (
            mean(control_percentages)
            if control_percentages else None
        )

        treated_mean = (
            mean(treated_percentages)
            if treated_percentages else None
        )

        # Calculate treated minus control difference
        if control_mean is not None and treated_mean is not None:
            difference = treated_mean - control_mean
        else:
            difference = None

        # Only perform the test if both groups have enough samples
        if len(control_percentages) < 2 or len(treated_percentages) < 2:
            statistical_result = {
                "statistic": None,
                "p_value": None
            }

        else:
            statistical_result = calculate_statistics(
                control_percentages,
                treated_percentages
            )

        statistic = statistical_result["statistic"]
        p_value = statistical_result["p_value"]

        # Handle unavailable or non-finite statistical results
        if (
            statistic is None
            or p_value is None
            or not isfinite(statistic)
            or not isfinite(p_value)
        ):
            statistic = None
            p_value = None

        # Create the exact dictionary structure required
        result = {
            "position": position,
            "control_n": len(control_percentages),
            "treated_n": len(treated_percentages),
            "control_mean": control_mean,
            "treated_mean": treated_mean,
            "difference": difference,
            "statistic": statistic,
            "p_value": p_value
        }

        results.append(result)

        # Calculate degrees of freedom for scientific reporting
        df = welch_df(
            control_percentages,
            treated_percentages
        )

        # Print the results using the formatting helper
        print_position_result(
            result, df, control_group, treated_group
        )

    return results
````
